```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os 
```

```python
def cropped_black_borders(img:np.ndarray,thr=12,pad=0):
    #thr param : thresh değerlerini temsil ediyor arkaplan siyah renginde olduğu için 0 a çok yakın bir değer verdikk pad param:
    #padding'i temsil ediyor görüntüyü arkaplandan ayırt ettikten sonra ....

    #ilk olarak eğer görüntü renkli bir görüntü ise yani 3 kanallı şse bunun kontorlünü sağlıyoruz:
    if img.ndim==3:
        gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    else:
        gray = img.copy()
        

    #görüntü tek kannalı hale çevirdikten sonra arka plandaki siyah piksellerden ayıtr etmek için maske çıkarıyoruz
    
    mask=(gray>thr).astype(np.uint8)*255
    kernel=np.ones((5,5),dtype=np.uint8)
    
    thresh= cv2.morphologyEx(mask,cv2.MORPH_CLOSE,kernel,iterations=2)
    
    contour,_=cv2.findContours(mask,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
    
    c= max(contour,key=cv2.contourArea)
    
    x,y,w,h=cv2.boundingRect(c)
    
    H,W=img.shape[:2]
    
    x1=max(0,x-pad)
    y1=max(0,y-pad)
    x2=min(W,x+w+pad)
    y2=min(H,y+h+pad)
    
    return img[y1:y2,x1:x2]
```

```python
my_bgr_img=cv2.imread("./yeni1.png")

my_cropped_img=cropped_black_borders(my_bgr_img)
cv2.imshow("cropped image",my_cropped_img)
cv2.waitKey(0)
cv2.destroyAllwindows()
```
