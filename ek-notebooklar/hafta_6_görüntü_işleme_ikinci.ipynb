```python
import cv2
import numpy as np
```

```python
model_pathh='haarcascade_frontalface_default.xml'
face_cascades=cv2.CascadeClassifier(cv2.data.haarcascades + model_pathh)
```

```python
my_bgr_img=cv2.imread("./deneme.jpg")
my_gray_img=cv2.imread("./deneme.jpg",0)
```

```python
faces= face_cascades.detectMultiScale(my_gray_img,scaleFactor=1.1,minNeighbors=5,minSize=(5,5))
sayac=0

for(x,y,w,h) in faces:
    bos_list=cv2.rectangle(my_bgr_img,(x,y),(x+w,y+h),(0,255,0),2)
    sayac=sayac+1



print(sayac)
cv2.imshow("Futbolcular",my_bgr_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
