```python
# pythonda 2 adet cv2,resize() fonksiyonu yazınız
# bu fonksiyonLAR 1 adet parametre alacak. o da görüntünün kendisi olacak görüntüyü büyütürken de genişlik ve yüksekliğinin
#%25 'i kadar olacaktır her iki fonksiyon da geriye işlenmiş görüntünün kendisini döndüreceketir

# cv2.resize() kullanımı
#cv2.resize(original_img,(genişlik,yükseklik),cv2.INTER_LANCZOS4)

def img_zoom_in(img_org): 
    #görüntünün satırının (yüksekliğinin) %25 'ini hesapladık
    new_height = int(img_org.shape[0]) + int(img_org.shape[0]*0.25)
    new_weight = int(img_org.shape[1]) + int(img_org.shape[1]*0.25)

    new_img =cv2.resize(img_org,(new_weight,new_height),cv2.INTER_LANCZOS4)
           
    return new_img




def img_zoom_out(img_org): 
    #görüntünün satırının (yüksekliğinin) %25 'ini hesapladık
    new_height = int(img_org.shape[0]) - int(img_org.shape[0]*0.25)
    new_weight = int(img_org.shape[1]) - int(img_org.shape[1]*0.25)

    new_img =cv2.resize(img_org,(new_weight,new_height),cv2.INTER_LANCZOS4)
           
    return new_img
```

```python
my_rgb_img = cv2.cvtColor(my_img, cv2.COLOR_BGR2RGB)
```

```python
new_big_img = img_zoom_in(my_rgb_img)
new_small_img = img_zoom_out(my_rgb_img)

print(f" orijinal görüntü = {my_rgb_img.shape[:2]}")
print(f" büyütülmüş görüntü = {new_big_img.shape[:2]}")
print(f" küçütülmüş görüntü = {new_small_img.shape[:2]}")
```
