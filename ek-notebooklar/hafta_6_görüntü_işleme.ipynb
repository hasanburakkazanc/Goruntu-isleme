```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
my_bgr_img=cv2.imread("./bonus.jpg")
my_gray_img=cv2.cvtColor(my_bgr_img,cv2.COLOR_BGR2GRAY)
```

```python
type(my_bgr_img)
```

```python
type(my_gray_img)
```

```python
#ilk iş olarak gri seviyedeki görüntünün eşiğini hesaplayarak binaty hale getiriyoruz bu sayede kenar tespitini en başarılı şekile
# yapıp bu kenarların birleşmesi sonucu kontoru bulmayı hedefliyoruz

ret,thresh=cv2.threshold(my_gray_img,127,255,cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
thresh=cv2.morphologyEx(thresh,cv2.MORPH_CLOSE,np.ones((3,3)),iterations=2)

# RET olarak thresh değeri dönüyüor yani 127 trhesh değişkeni de şikelenmiş görüntüü saklıyor
# eşiklenmiş görüntüyü elde ettikten sonra kontrulraı bulmaya başlıyoruz kontur bulma işleminde de 2 adet değer dönüyor konturların hiyareişisini
# [anne /çocuk ilişkisi) ve konturların listesi. bize lazım olan değer konturların listesi yani dönen ikinci değişken

contour,hierarchy = cv2.findContours(thresh,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

cv2.drawContours(my_bgr_img,contour,-1,(0,255,0),2)


cv2.imshow("Image's Contour",my_bgr_img)
cv2.imshow("thresh Image ",thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
